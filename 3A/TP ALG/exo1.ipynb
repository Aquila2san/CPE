```python
import numpy as np
import numpy.linalg as alg
```

## Question 1

```python
A = np.array([
    [1, -2, 3, 1],
    [2, 5, -1, 3],
    [4, 1, 0, 2],
    [0, -2, 3, 1]
])

print("1. Matrice A")
print(A, '\n')
```

## Question 2

```python
print("2. Slicing")
print("3ème colonne :")
print(A[:, 2], '\n')

print("4ème ligne :")
print(A[3, :], '\n')
```

## Question 3

```python
A[:, 2] = 1 
print("3. Matrice A modifiée")
print(A, '\n')
```

## Question 4

```python
det_A = alg.det(A)
trace_A = alg.trace(A)

print("4. Trace et Déterminant")
print(f"Déterminant : {det_A}")
print(f"Trace : {trace_A}\n")
```

## Question 5

```python
b = np.array([[1], [2], [-1], [3]])
A_inv = alg.inv(A)
x = A_inv @ b

print("5. Résolution Ax=b et Inverse")
print("Solution x :\n", x)
print("Inverse de A :\n", A_inv, '\n')
```
