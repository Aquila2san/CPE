```python
import numpy as np
import matplotlib.pyplot as plt
import numpy.linalg as alg

# Taille de la matrice
m, n = 20, 10

# Construction de la matrice binaire
A = np.random.randint(0, 2, (m, n))

# Affichage de l'image correspondant à la matrice A (0->noir, 1->blanc)
plt.figure(1,figsize=(12, 4))
plt.subplot(131)
plt.imshow(A, cmap='gray')
plt.title("a. Image d'origine (A)", fontsize=14)

# Décomposition en valeurs singulières de A (utiliser numpy.linalg.svd)
U, s, Vt = alg.svd(A)

# Affichage des valeurs singulières
plt.subplot(132)
plt.plot(s, '*k-')
plt.grid(True)
plt.title("b. Valeurs singulières", fontsize=14)

# Nombre de valeurs singulières non nulles
r = alg.matrix_rank(A)
print(f"Rang de la matrice (valeurs singulières non nulles) : {r}")
print(f"Nombre total de valeurs singulières : {len(s)}")

# Reconstruction de la matrice avec les valeurs singulières principales
B = np.zeros((m, n))   # initialisation
k_limite = len(s) - 3
for k in range(k_limite):
    B += U[:, :k] @ np.diag(s[:k]) @ Vt[:k, :]

# Affichage de l'image correspondant à la matrice B
plt.subplot(133)
plt.imshow(B, cmap='gray')
plt.title("c. Image reconstruite (B)", fontsize=14)
plt.show()


     
```

```python
A = plt.imread("einstein.jpg")
A = A[:,:,0]

plt.figure(1,figsize=(12, 4))
plt.subplot(131)
plt.imshow(A, cmap='gray')
plt.title("a. Image d'origine (A)", fontsize=14)

U, s, Vt = alg.svd(A)

plt.subplot(132)
plt.plot(s, '*k-')
plt.grid(True)
plt.title("b. Valeurs singulières", fontsize=14)

B5 = np.zeros((len(U), len(Vt[0]))) 
for k in range(1,6):
    B5 += U[:, :k] @ np.diag(s[:k]) @ Vt[:k, :]

B40 = np.zeros((len(U), len(Vt[0]))) 
for k in range(1,41):
    B40 += U[:, :k] @ np.diag(s[:k]) @ Vt[:k, :]

B100 = np.zeros((len(U), len(Vt[0]))) 
for k in range(1,101):
    B100 += U[:, :k] @ np.diag(s[:k]) @ Vt[:k, :]

B200 = np.zeros((len(U), len(Vt[0]))) 
for k in range(1,201):
    B200 += U[:, :k] @ np.diag(s[:k]) @ Vt[:k, :]


plt.subplot(133)
plt.imshow(B5, cmap='gray')
plt.title("c. Image reconstruite (B5)", fontsize=14)
plt.show()

plt.figure(1,figsize=(12, 4))
plt.imshow(B40, cmap='gray')
plt.title("c. Image reconstruite (B40)", fontsize=14)
plt.show()

plt.figure(1,figsize=(12, 4))
plt.imshow(B100, cmap='gray')
plt.title("c. Image reconstruite (B100)", fontsize=14)
plt.show()

plt.figure(1,figsize=(12, 4))
plt.imshow(B200, cmap='gray')
plt.title("c. Image reconstruite (B200)", fontsize=14)
plt.show()

m, n = A.shape
k_range = np.arange(1, min(m, n) + 1)

tau = 1 - (k_range * (m + n + 1)) / (m * n)

plt.figure(figsize=(8, 5))
plt.plot(k_range, tau)
plt.xlabel("Nombre de valeurs singulières conservées (k)")
plt.ylabel("Taux de compression t")
plt.title("Taux de compression théorique en fonction de k")
plt.grid(True)
plt.show()
```
