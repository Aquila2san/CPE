```python
import numpy as np
import matplotlib.pyplot as plt
import numpy.linalg as alg

# Définition du motif de base
F0 = 2 * np.array([[-0.5, -0.5, -5, -3, -10, -8, -9, -6, -6, -2, -5, -2, 0, 2, 5, 2, 6, 6, 9, 8, 10, 3, 5, 0.5, 0.5],  # abscisses
                   [0, 10, 9, 12, 17, 17, 20, 20, 22, 17, 27, 25, 30, 25, 27, 17, 22, 20, 20, 17, 17, 12, 9, 10, 0]])  # ordonnées
m, n = F0.shape   # dimension de la matrice F0

# Taille du cadre d'affichage
L = 200
cadre = [-L, L, -L, L]

# F1 l'image par translation de F0
u = np.array([[-80], [60]])
F1 = F0 + u

# F2 le symétrique de F1
N = np.array([[np.sqrt(2)/2], [np.sqrt(2)/2]])
I = np.eye(2)
S_mat = 2 * (N @ N.T) - I
F2 = S_mat @ F1

# F3 l'image de F2 par rotation
theta = -2 * np.pi / 3
R_mat = np.array([[np.cos(theta), -np.sin(theta)],
                  [np.sin(theta), np.cos(theta)]])
F3 = R_mat @ F2

# Affichage des motifs
plt.figure(1, figsize=(8, 8))
plt.fill(F0[0, :], F0[1, :], c=[0,0,0])   # [0,0,0]=noir
plt.fill(F1[0, :], F1[1, :], c='r') 
plt.fill(F2[0, :], F2[1, :], c='g') 
plt.fill(F3[0, :], F3[1, :], c='b')
plt.axis('equal')
plt.axis(cadre)

# Axe de symétrie
t = np.linspace(-1000, 1000, 100)
plt.plot(t*N[0], t*N[1], 'g-.')

# Cercle de centre O 
# Calcul du centre de gravité G2 de la feuille F2
G2 = np.mean(F2, axis=1)
r = alg.norm(G2)

# Tracé du cercle
angles = np.linspace(0, 2*np.pi, 200)
plt.plot(r * np.cos(angles), r * np.sin(angles), 'b:')
```
