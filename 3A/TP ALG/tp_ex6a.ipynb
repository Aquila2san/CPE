```python
import numpy as np
import matplotlib.pyplot as plt

# Partie A : Descente de gradient pour une fonction polynomiale simple

# Définition de la fonction f
def f(x,a,b,c):
    return a * x**2 + b * x + c

xmin, xmax = -2, 3
x = np.linspace(xmin, xmax, 1000)

# Paramètres de la fonction
a, b, c = 2, -3, 1
y = f(x, a, b, c)                                     # Valeurs de y pour chaque x
x_th = -b / (2 * a)                                   # Minimum théorique

plt.figure(1)
plt.plot(x, y)
plt.grid(True)

def df(x):
    return 4 * x - 3

x0 = 3          
lam = 0.1      
epsilon = 1e-3  
x = x0
x_historique = [x0] 
iterations = 0

while abs(df(x)) >= epsilon:
    grad = df(x)
    x = x - lam * grad
    x_historique.append(x)
    iterations += 1
    if iterations > 1000:
        break

print ("Le nombre d'itérations nécéssaires pur obtenir le minimum vaut :", iterations)

x_grad = x_historique[-1]

delta = np.abs((x_th - x_grad) / x_th)
print ("L'erreur relative vaut :", delta)
```
