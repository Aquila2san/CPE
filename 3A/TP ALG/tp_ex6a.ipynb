```python
import numpy as np
import matplotlib.pyplot as plt

# Partie A : Descente de gradient pour une fonction polynomiale simple

# Définition de la fonction f
def f(x,a,b,c):
    return a * x**2 + b * x + c

xmin, xmax = -2, 3
x = np.linspace(xmin, xmax, 1000)

# Paramètres de la fonction
a, b, c = 2, -3, 1
y = f(x, a, b, c)                                     # Valeurs de y pour chaque x
x_th = -b / (2 * a)                                   # Minimum théorique

plt.figure(1)
plt.plot(x, y)
plt.grid(True)
```
