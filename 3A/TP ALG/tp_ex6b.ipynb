```python
import numpy as np
import pandas as pd
import plotly.express as px

# Définir l'équation de la surface z = f(x, y)
def f(x, y):
    return x - y + 2 * x**2 + 2 * x * y + y**2

# Créer une grille de points pour x et y
x_vals = np.linspace(-5, 5, 100)
y_vals = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x_vals, y_vals)
Z = f(X, Y)

# Organiser les données de la surface dans un DataFrame
surface_data = pd.DataFrame({
    'x': X.ravel(),
    'y': Y.ravel(),
    'z': Z.ravel()
})

# Ajouter des points supplémentaires
points_data = pd.DataFrame({
    'x': [-1, 1, 4],
    'y': [-3, 1, 2],
    'z': [1, 4, 2]
})

# Tracer la surface
fig = px.scatter_3d(
    surface_data,
    x='x',
    y='y',
    z='z',
    opacity=0.6,
    color='z',
    color_continuous_scale='Viridis',
    title="Surface et points supplémentaires"
)

# Ajouter les points supplémentaires
fig.add_scatter3d(
    x=points_data['x'],
    y=points_data['y'],
    z=points_data['z'],
    mode='markers',
    marker=dict(size=8, color='red'),
    name='Points supplémentaires'
)

# Configurer l'affichage
fig.update_layout(
    scene=dict(
        xaxis_title="X",
        yaxis_title="Y",
        zaxis_title="Z"
    ),
    width=800,
    height=800
)

# Afficher la figure
fig.show()

def grad_f(x, y):
    df_dx = 4*x + 2*y + 1
    df_dy = 2*x + 2*y - 1
    return np.array([df_dx, df_dy])

x_start = np.array([3.0, 4.0])
lam = 0.1
epsilon = 1e-2
xk = x_start
chemin = [xk]
iterations = 0

while np.linalg.norm(grad_f(xk[0], xk[1])) >= epsilon:
    grad = grad_f(xk[0], xk[1])
    xk = xk - lam * grad
    chemin.append(xk)
    iterations += 1
    if iterations > 1000: 
        break

xk_final = chemin[-1]
x_th = np.array([-1.0, 1.5])
erreur = np.linalg.norm(x_th - xk_final) / np.linalg.norm(x_th)

print("Minimum trouvé :", xk_final)
print("Itérations :", iterations)
print("Erreur relative :", erreur)
```
