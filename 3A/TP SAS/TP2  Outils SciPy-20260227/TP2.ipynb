```python
import numpy as np
from scipy import signal
from msicpe.sas import pplane, zplane, pdiagrammes, zdiagrammes
import plotly.express as px
import pandas as pd
from systems_tp import systems
```

```python
# Système analogique (à temps continu)
b_a, a_a = np.array([10000]), np.array([1., 26.13, 341.42, 2613.13, 10000])
system_a = signal.TransferFunction(b_a, a_a) 
# Donne les coefficients au numérateur et au dénumérateur du système
print(system_a.to_tf())
# Donne les matrices de réprésentation d'état du système
print(system_a.to_ss())
# Donne la forme factorisée (zéros, pôles, gain)
print(system_a.to_zpk())
# Tracé du diagramme Pôles-Zéros pour analyse de la stabilité
pplane(system_a)
# Tracés des diagrammes de Bode, Black-Nichols et Nyquist pour l'analyse fréquentielle
pdiagrammes(system_a)
# Tracé de la réponse indicielle
t_step, y_step = signal.step(system_a)
fig = px.line(x=t_step,y=y_step,
    labels={
        "x": "Temps (s)",
        "y": "Amplitude"
    },
    title="Réponse indicielle"
)
fig.update_layout(
    width=500,
    height=500
)
fig.show()
```

```python
# Système numérique (à temps discret)
b_n, a_n = np.array([0.018, 0.070, 0.106, 0.070, 0.018]), np.array([1.0, -1.609,  1.315, -0.504,  0.080])
system_n = signal.TransferFunction(b_n, a_n, dt=0.1)
# Donne les coefficients au numérateur et au dénumérateur du système
print(system_n.to_tf())
# Donne les matrices de réprésentation d'état du système
print(system_n.to_ss())
# Donne la forme factorisée (zéros, pôles, gain)
print(system_n.to_zpk())
# Tracé du diagramme Pôles-Zéros dans le plan complexe Z
zplane(system_n)
# Tracés des diagrammes fréquentiels discrets
zdiagrammes(system_n)
# Tracé de la réponse indicielle discrète
t, y = signal.dstep(system_n)
y = np.squeeze(y)
fig = px.scatter(
    x=t,
    y=y,
    labels={
        "x": "n (échantillons)",
        "y": "Amplitude"
    },
    title="Réponse indicielle discrète"
)

fig.update_layout(
    width=500,
    height=500
)

fig.show()
```
