### 1. Définition d'un spectrogramme

Un spectrogramme réalise l'analyse spectrale d'un signal $s(t)$ vu au travers d'une fenêtre d'observation $h(t)$ glissante. Il décrit l'évolution au cours du temps de l'énergie du signal en fonction de la fréquence. Son expression mathématique est :

$$S(t,\nu)=\left|\int_{-\infty}^{\infty}s(\theta)h(t-\theta)e^{-2i\pi\nu\theta}d\theta\right|^{2}$$


---

### 2. Principe de la méthode des fenêtres (Filtres RIF)

La méthode des fenêtres permet de synthétiser des filtres à Réponse Impulsionnelle Finie (RIF). 

* **Troncature :** La réponse impulsionnelle d'un filtre idéal (infinie) est tronquée pour obtenir un nombre fini de coefficients $M$.
* **Phénomène de Gibbs :** Une troncature brutale équivaut à l'utilisation d'une fenêtre rectangulaire, provoquant des oscillations dans les bandes passantes et atténuées.
* **Fenêtrage :** La réponse tronquée est multipliée par une fenêtre d'apodisation (ex: Hamming, Hann, Blackman) pour atténuer ces oscillations aux extrémités.

---

### 3. Paramètres de `scipy.signal.firwin`

La fonction `scipy.signal.firwin` synthétise un filtre RIF.

| Paramètre | Type | Rôle |
| :--- | :--- | :--- |
| `numtaps` | Entier | Longueur $M$ du filtre (nombre de coefficients). |
| `cutoff` | Flottant/Tableau | Fréquence(s) de coupure du filtre. |
| `window` | Chaîne | Fenêtre à appliquer ('boxcar', 'hamming', 'hann', 'blackman'). |
| `pass_zero` | Booléen/Chaîne | Nature du filtre (passe-bas, passe-haut, etc.). |
| `fs` | Flottant | Fréquence d'échantillonnage du signal. |

**Sortie :** Tableau unidimensionnel des coefficients de la réponse impulsionnelle.

---

### 4. Principe de la synthèse d'un filtre RII par transformée bilinéaire

La conception d'un filtre à Réponse Impulsionnelle Infinie (RII) numérique s'effectue à partir d'un filtre analogique.

1. **Synthèse analogique :** Conception dans le domaine continu (plan de Laplace $p$).
2. **Discrétisation :** La transformée bilinéaire convertit le filtre analogique en filtre numérique discret (plan $z$). Elle mappe le demi-plan gauche $\mathbb{R}(p)<0$ à l'intérieur du cercle unité $|z|<1$. Cela garantit la stabilité mais induit une distorsion non linéaire de l'axe des fréquences.

---

### 5. Paramètres de `scipy.signal.ellip` et `scipy.signal.bilinear`

**`scipy.signal.ellip`** (Génération du filtre analogique)

| Paramètre | Rôle |
| :--- | :--- |
| `N` | Ordre du filtre. |
| `rp` | Ondulation maximale dans la bande passante (dB). |
| `rs` | Atténuation minimale dans la bande atténuée (dB). |
| `Wn` | Fréquence(s) critique(s). |
| `analog` | Sur `True` pour générer un filtre analogique. |

**`scipy.signal.bilinear`** (Application de la transformée bilinéaire)

| Paramètre | Rôle |
| :--- | :--- |
| `b`, `a` | Coefficients du filtre analogique. |
| `fs` | Fréquence d'échantillonnage. |

```python
import numpy as np 
import scipy.signal as signal 
import plotly.express as px 
import pandas as pd # pour les dataframes 
import msicpe.tns as tns 
import sounddevice as sd 
import scipy.io.wavfile as wv
```

# 1 Analyse du signal et définition du gabarit de filtre idéal

```python
Fs,x=wv.read('Signaux/signal4.wav') 
x = x/np.max(np.abs(x)) # Normalisation du signal de départ 
sd.play(x,Fs) # attention à baisser le son à la première écoute
print(Fs)
```

```python
# Fonction pour afficher le spectrogramme 
def plot_spectrogram(x, Fs, title): 
    # Paramètres du spectrogramme 
    size_win = 512 
    win = signal.windows.hann(size_win) 
    hop = size_win//4 
    SFT = signal.ShortTimeFFT(win=win, hop=hop, fs=Fs, mfft=size_win*4) 
    # Calcul du spectrogramme 
    spectrogram = SFT.stft(x) 
    lims = SFT.extent(x.size) 
    # Tracé du spectrogramme 
    t_spec = np.arange(lims[0], lims[1], SFT.delta_t) 
    px.imshow(20*np.log10(np.abs(spectrogram)), 
              x = t_spec, y = SFT.f, origin='lower', aspect='auto', labels={'x': 'Temps (s)', 'y': 'Fréquence (Hz)', 'color': 'Magnitude'}, zmin=-70, title=title).show()
    
plot_spectrogram(x,Fs, 'Spectrogramme du signal 4')
```

 On remarque que entre 0 et 3000 Hz on a le signal audio utile qui nous intéressent et que entre 3000 et 6400 Hz on a une large bande de magnitude importante qui correspond au bruit. Ainsi pour conserver uniquement le signal audio utile il faut utiliser un filtre passe-bas à la fréquence de coupure fc = 3000 Hz

# 2 Synthèse de filtres RIF

```python
# Définition des constantes
fc = 3000
M = 151

# Synthèse du filtre RIF
h_rect = signal.firwin(numtaps=M, cutoff=fc, window='boxcar', pass_zero='lowpass', fs=Fs)

# Calcul de la réponse en fréquence
f, h_freq = signal.freqz(h_rect, worN=2048, fs=Fs)
magnitude_db = 20 * np.log10(np.abs(h_freq))
phase = np.unwrap(np.angle(h_freq))

# Affichage
px.line(x=np.arange(M), y=h_rect, labels={"x":"Echantillons", "y":"Amplitude"},title='Réponse impulsionnelle').show()
px.line(x=f, y=magnitude_db, labels={"x":"Fréquence (Hz)", "y":"Amplitude (dB)"},title='Module du gain complexe').show()
px.line(x=f, y=phase, labels={"x":"Fréquence (Hz)", "y":"Phase (en rad)"},title='Phase du gain complexe').show()

tns.zplane(np.roots(h_rect), [])
```

```python
h_blackman = signal.firwin(numtaps=M, cutoff=fc, window='blackman', pass_zero='lowpass', fs=Fs)

f, H_blackman = signal.freqz(h_blackman, worN=2048, fs=Fs)
mag_blackman_db = 20 * np.log10(np.abs(H_blackman))
phase_blackman = np.unwrap(np.angle(H_blackman))

px.line(x=np.arange(M), y=h_blackman, labels={"x":"Echantillons", "y":"Amplitude"},title='Réponse impulsionnelle').show()
px.line(x=f, y=mag_blackman_db, labels={"x":"Fréquence (Hz)", "y":"Amplitude (dB)"},title='Module du gain complexe').show()
px.line(x=f, y=phase_blackman, labels={"x":"Fréquence (Hz)", "y":"Phase (en rad)"},title='Phase du gain complexe').show()

tns.zplane(np.roots(h_blackman), [])
```

```python
df = pd.DataFrame({
    'Fréquence (Hz)': np.concatenate([f, f]),
    'Amplitude linéaire': np.concatenate([np.abs(h_freq), np.abs(H_blackman)]),
    'Filtre': ['Rectangulaire'] * len(f) + ['Blackman'] * len(f)
})

px.line(df, x='Fréquence (Hz)', y='Amplitude linéaire', color='Filtre', title="Modules du gain complexe ").show()
```

```python
# Filtrage avec lfilter
y_rect = signal.lfilter(h_rect, 1.0, x)
y_black = signal.lfilter(h_blackman, 1.0, x)

# Normalisation des signaux filtrés
y_rect = y_rect / np.max(np.abs(y_rect))
y_black = y_black / np.max(np.abs(y_black))

# Écoute 
#sd.play(y_rect, Fs)
#sd.play(y_black, Fs)


# Tracé des spectrogrammes
plot_spectrogram(y_rect, Fs, "Spectrogramme - Fenêtre Rectangulaire")
plot_spectrogram(y_black, Fs, "Spectrogramme - Fenêtre Blackman")
```
